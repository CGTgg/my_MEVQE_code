```python
def calculate_fi(sequence):
    total_abs_diff = 0.0
    n = len(sequence)
    
    # 计算相邻元素的绝对差值之和
    for i in range(1, n):
        diff = abs(sequence[i] - sequence[i-1])
        total_abs_diff += diff
    
    # FI = 绝对差之和 / 数据总长度
    fi = total_abs_diff / n
    return fi

# 输入数据
L1 = processed_L7

# 计算波动指数
fi_result = calculate_fi(L1)

# 输出结果（保留5位小数）
print(f"波动指数 FI = {fi_result:.5f}")
```
