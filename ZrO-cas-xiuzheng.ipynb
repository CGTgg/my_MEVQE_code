```python
from pyscf import gto, scf, mcscf
import numpy as np
```

```python
# 步骤1：分子结构与HF计算
# ====================
#m-ZrO2
mol1 = gto.M(
    atom='Zr 1.2397611467218508 0.22042881 1.102028681533527; O 4.657048362037544 4.36031376 0.8211243583071115;O 1.4584165528787967 3.8836934000000003 5.145324267709785',
    basis='gth-dzvp-molopt-sr',
    spin=0,
    verbose=4
)
mf1 = scf.RHF(mol1).run()
```

```python
# 步骤2：初始活性空间上的计算
# ====================
# 初始活性空间：CAS(3,4) 3轨道4电子
ncas_initial = 3
nelecas_initial = (2, 2)  # 4电子

# 运行CASSCF
mc_initial1 = mcscf.CASSCF(mf1, ncas_initial, nelecas_initial)
mc_initial1.kernel()
e_initial1 = mc_initial1.e_tot 
```

```python
# 步骤3：扩展活性空间计算（动态关联补偿）
# ====================
# 子步骤3.1：自动选择扩展轨道
ncas_extended = 8  # 扩展至6轨道
nelecas_initial = (4, 4)  # 4电子


# 子步骤3.2：运行扩展CASSCF
mc_extended1 = mcscf.CASSCF(mf1, ncas_extended, nelecas_initial)

# 优化扩展活性空间
#mc_extended.internal_rotation = True  # 启用轨道优化
mc_extended1.kernel()
```

```python
# 步骤4：能量修正计算
# ====================
delta_e1 = mc_extended1.e_tot - mc_initial1.e_tot

print(f"""
[能量修正报告] m-ZrO2分子CASSCF补偿结果
---------------------------------------
初始活性空间 (CAS({ncas_initial},4)) : {mc_initial1.e_tot:.6f} Hartree
扩展活性空间 (CAS(8,8))        : {mc_extended1.e_tot:.6f} Hartree
动态关联修正 ΔE               : {delta_e1:.6f} Hartree
---------------------------------------
""")
```
