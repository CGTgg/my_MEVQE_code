```python
import csv
import time
import itertools
from pathlib import Path
from collections import OrderedDict

import numpy as np
import matplotlib.pyplot as plt
import mindspore as ms
from mindspore import context, nn, Parameter, Tensor, ops
from mindspore.common.initializer import initializer

from openfermion.chem import MolecularData
from openfermionpyscf import run_pyscf
from openfermion.transforms import get_fermion_operator, jordan_wigner
from openfermion.linalg import get_ground_state, get_sparse_operator

from mindquantum.core.gates import X
from mindquantum.core.circuit import Circuit
from mindquantum.core.operators import Hamiltonian, FermionOperator
from mindquantum.simulator import Simulator
from mindquantum.algorithm.nisq import Transform
from mindquantum.framework import MQEncoderOnlyOps
from mindquantum.third_party.interaction_operator import InteractionOperator
from mindquantum.core.operators import down_index, up_index
from mindquantum.utils.type_value_check import _require_package

# ============================================================
# 0. 环境设置
# ============================================================
ms.set_context(mode=ms.PYNATIVE_MODE, device_target="CPU")


# ============================================================
# 1. UCCSD 生成相关函数
# ============================================================
def _para_uccsd_singlet_generator(mol, threshold=0, n_qubits=0, n_electrons=0):
    """Generate a uccsd quantum circuit with parameter names."""
    if n_qubits == 0:
        n_qubits = mol.n_qubits
        n_electrons = mol.n_electrons
    params = {}
    if n_qubits % 2 != 0:
        raise ValueError('The total number of spin-orbitals should be even.')
    out = []
    n_spatial_orbitals = n_qubits // 2
    n_occupied = int(np.ceil(n_electrons / 2))
    n_virtual = n_spatial_orbitals - n_occupied

    n_single_amplitudes = n_occupied * n_virtual
    spin_index_functions = [up_index, down_index]

    # Single and double excitations from one spatial occupied-virtual pair
    for i, (p, q) in enumerate(itertools.product(range(n_virtual), range(n_occupied))):
        virtual_spatial = n_occupied + p
        occupied_spatial = q
        virtual_up = virtual_spatial * 2
        occupied_up = occupied_spatial * 2
        virtual_down = virtual_spatial * 2 + 1
        occupied_down = occupied_spatial * 2 + 1
        single_amps = mol.ccsd_single_amps[virtual_up, occupied_up]
        double1_amps = mol.ccsd_double_amps[virtual_up, occupied_up, virtual_down, occupied_down]
        single_amps_name = 'p' + str(i)
        double1_amps_name = 'p' + str(i + n_single_amplitudes)

        for spin in range(2):
            this_index = spin_index_functions[spin]
            other_index = spin_index_functions[1 - spin]
            virtual_this = this_index(virtual_spatial)
            virtual_other = other_index(virtual_spatial)
            occupied_this = this_index(occupied_spatial)
            occupied_other = other_index(occupied_spatial)

            if abs(single_amps) > threshold:
                params[single_amps_name] = single_amps
                fermion_ops1 = FermionOperator(((occupied_this, 1), (virtual_this, 0)), 1)
                fermion_ops2 = FermionOperator(((virtual_this, 1), (occupied_this, 0)), 1)
                out.append([fermion_ops1 - fermion_ops2, single_amps_name])

            if abs(double1_amps) > threshold:
                params[double1_amps_name] = double1_amps
                fermion_ops1 = FermionOperator(
                    ((virtual_this, 1), (occupied_this, 0), (virtual_other, 1), (occupied_other, 0)), 1
                )
                fermion_ops2 = FermionOperator(
                    ((occupied_other, 1), (virtual_other, 0), (occupied_this, 1), (virtual_this, 0)), 1
                )
                out.append([fermion_ops1 - fermion_ops2, double1_amps_name])

    # Double excitations from two spatial occupied-virtual pairs
    for i, ((p, q), (r, s)) in enumerate(
        itertools.combinations(itertools.product(range(n_virtual), range(n_occupied)), 2)
    ):
        virtual_spatial_1 = n_occupied + p
        occupied_spatial_1 = q
        virtual_spatial_2 = n_occupied + r
        occupied_spatial_2 = s

        virtual_1_up = virtual_spatial_1 * 2
        occupied_1_up = occupied_spatial_1 * 2
        virtual_2_up = virtual_spatial_2 * 2 + 1
        occupied_2_up = occupied_spatial_2 * 2 + 1

        double2_amps = mol.ccsd_double_amps[virtual_1_up, occupied_1_up, virtual_2_up, occupied_2_up]
        double2_amps_name = 'p' + str(i + 2 * n_single_amplitudes)

        for (spin_a, spin_b) in itertools.product(range(2), repeat=2):
            index_a = spin_index_functions[spin_a]
            index_b = spin_index_functions[spin_b]
            virtual_1_a = index_a(virtual_spatial_1)
            occupied_1_a = index_a(occupied_spatial_1)
            virtual_2_b = index_b(virtual_spatial_2)
            occupied_2_b = index_b(occupied_spatial_2)
            if abs(double2_amps) > threshold:
                params[double2_amps_name] = double2_amps
                fermion_ops1 = FermionOperator(
                    ((virtual_1_a, 1), (occupied_1_a, 0), (virtual_2_b, 1), (occupied_2_b, 0)), 1
                )
                fermion_ops2 = FermionOperator(
                    ((occupied_2_b, 1), (virtual_2_b, 0), (occupied_1_a, 1), (virtual_1_a, 0)), 1
                )
                out.append([fermion_ops1 - fermion_ops2, double2_amps_name])

    return out, params


def _transform2pauli(fermion_ansatz):
    """Transform fermion ansatz to pauli ansatz."""
    pauli_ansatz = []
    for fermion_op, param_name in fermion_ansatz:
        qubit_op = Transform(fermion_op).jordan_wigner()
        pauli_ansatz.append([qubit_op, param_name])
    return pauli_ansatz


def _pauli2circuit(pauli_ansatz):
    """Transform pauli ansatz to mindquantum circuit."""
    circuit = Circuit()
    for pauli_op, param_name in pauli_ansatz:
        for term, coeff in pauli_op.terms.items():
            if abs(coeff) < 1e-12:
                continue
            sub_circuit = Circuit()
            for qubit_idx, pauli_gate in term:
                if pauli_gate == 'X':
                    sub_circuit += X.on(qubit_idx)
                elif pauli_gate == 'Y':
                    from mindquantum.core.gates import Y
                    sub_circuit += Y.on(qubit_idx)
                elif pauli_gate == 'Z':
                    from mindquantum.core.gates import Z
                    sub_circuit += Z.on(qubit_idx)
            from mindquantum.core.operators import TimeEvolution
            te = TimeEvolution(sub_circuit.to_pauli_words()[0][0], {param_name: 1.0})
            circuit += te.circuit
    return circuit


def generate_uccsd(molecular, threshold=-1, active_space_start=None, active_space_stop=None):
    """Generate UCCSD circuit from molecular data."""
    try:
        from openfermion.chem import MolecularData
    except (ImportError, AttributeError):
        _require_package("openfermion", "1.5.0")

    if isinstance(molecular, str):
        mol = MolecularData(filename=molecular)
        mol.load()
    else:
        mol = molecular

    n_active_qubits = 2 * (active_space_stop - active_space_start)
    n_active_electrons = mol.n_electrons - 2 * active_space_start

    fermion_ansatz, parameters = _para_uccsd_singlet_generator(
        mol, threshold, n_qubits=n_active_qubits, n_electrons=n_active_electrons
    )
    pauli_ansatz = _transform2pauli(fermion_ansatz)
    uccsd_circuit = _pauli2circuit(pauli_ansatz)

    ham_of = mol.get_molecular_hamiltonian(
        list(range(active_space_start)),
        active_indices=range(active_space_start, active_space_stop)
    )
    inter_ops = InteractionOperator(*ham_of.n_body_tensors.values())
    ham_hiq = FermionOperator(inter_ops)
    qubit_hamiltonian = Transform(ham_hiq).jordan_wigner()
    qubit_hamiltonian = qubit_hamiltonian.compress()

    parameters_name = list(parameters.keys())
    initial_amplitudes = [parameters[i] for i in parameters_name]
    return uccsd_circuit, initial_amplitudes, parameters_name, qubit_hamiltonian, n_active_qubits, n_active_electrons


# ============================================================
# 2. 分子构建
# ============================================================
def get_mol(phase=1):
    if phase == 1:
        a, b, c = 5.15, 5.23, 5.33
        beta = 99.44 * np.pi / 180
        a_vec = np.array([a, 0, 0])
        b_vec = np.array([0, b, 0])
        c_vec = np.array([c * np.cos(beta), 0, c * np.sin(beta)])
        atoms = [
            ('Zr', [0.276309, 0.042147, 0.209598]),
            ('O',  [0.930791, 0.833712, 0.156172]),
            ('O',  [0.449303, 0.742580, 0.978604])
        ]
        description = "Monoclinic ZrO2"
    elif phase == 2:
        a = b = 3.60
        c = 5.23
        a_vec = np.array([a, 0, 0])
        b_vec = np.array([0, b, 0])
        c_vec = np.array([0, 0, c])
        atoms = [('Zr', [0, 0, 0]), ('O', [0, 0.5, 0.301775])]
        description = "Tetragonal ZrO2"
    elif phase == 3:
        a, c = 5.31, 3.2
        a_vec = np.array([a, 0, 0])
        b_vec = np.array([-a / 2, a * np.sqrt(3) / 2, 0])
        c_vec = np.array([0, 0, c])
        atoms = [('Zr', [0, 0, 0]), ('O', [1 / 3, 2 / 3, 0.25])]
        description = "Hexagonal ZrO"
    else:
        raise ValueError("Invalid phase. Choose 1, 2, or 3.")

    geometry = []
    for element, frac_coords in atoms:
        cart_coords = frac_coords[0] * a_vec + frac_coords[1] * b_vec + frac_coords[2] * c_vec
        geometry.append([element, list(cart_coords)])

    basis = 'gth-dzvp-molopt-sr'
    multiplicity = 1
    molecule = MolecularData(geometry, basis, multiplicity, description=description)

    try:
        molecule.load()
    except Exception:
        molecule = run_pyscf(molecule, run_scf=1, run_ccsd=1)
        molecule.save()
        molecule.load()

    print(f'{description} - HF energy: {molecule.hf_energy} Hartree')
    print(f'{description} - CCSD energy: {molecule.ccsd_energy} Hartree')
    return molecule


# ============================================================
# 3. 线路资源统计
# ============================================================
def _read_gate_attr(gate, attr, default=None):
    value = getattr(gate, attr, default)
    if callable(value):
        try:
            return value()
        except TypeError:
            return default
    return value


def circuit_resource_metrics(circuit):
    gate_count = 0
    parameterized_gate_count = 0
    gate_type_count = {}
    qubit_depth = {}

    for gate in circuit:
        gate_name = str(_read_gate_attr(gate, "name", gate.__class__.__name__))
        gate_name_upper = gate_name.upper()
        if "MEASURE" in gate_name_upper or "BARRIER" in gate_name_upper:
            continue

        obj_qubits = _read_gate_attr(gate, "obj_qubits", ()) or ()
        ctrl_qubits = _read_gate_attr(gate, "ctrl_qubits", ()) or ()
        try:
            touched_qubits = set(obj_qubits) | set(ctrl_qubits)
        except TypeError:
            touched_qubits = set()

        gate_count += 1
        gate_type_count[gate_name] = gate_type_count.get(gate_name, 0) + 1
        if bool(_read_gate_attr(gate, "parameterized", False)):
            parameterized_gate_count += 1
        if touched_qubits:
            current_layer = max((qubit_depth.get(q, 0) for q in touched_qubits), default=0)
            next_layer = current_layer + 1
            for q in touched_qubits:
                qubit_depth[q] = next_layer

    circuit_depth = max(qubit_depth.values(), default=0)
    return {
        "n_qubits": int(circuit.n_qubits),
        "n_parameters": len(circuit.params_name),
        "gate_count": gate_count,
        "parameterized_gate_count": parameterized_gate_count,
        "circuit_depth": circuit_depth,
        "gate_type_count": gate_type_count,
    }


# ============================================================
# 4. 运行时统计 & LSTM 网络定义
# ============================================================
class RuntimeStats:
    def __init__(self):
        self.qnn_forward_calls = 0
        self.qnn_forward_seconds = 0.0
        self.lstm_forward_seconds = 0.0


class TrainableLSTM(nn.Cell):
    def __init__(self, n_parameters):
        super().__init__()
        self.n_parameters = int(n_parameters)
        self.lstm = nn.LSTM(
            input_size=self.n_parameters,
            hidden_size=self.n_parameters,
            num_layers=1,
            has_bias=True,
            batch_first=False,
            bidirectional=False,
        )
        self.hidden_size = self.lstm.hidden_size
        self.batch_size = 1
        self.num_layers = 1
        self.hx = (
            Parameter(initializer("normal", [self.num_layers, self.batch_size, self.hidden_size], dtype=ms.float32), name="lstm_h"),
            Parameter(initializer("normal", [self.num_layers, self.batch_size, self.hidden_size], dtype=ms.float32), name="lstm_c"),
        )

    def construct(self, x):
        _, (hidden, _) = self.lstm(x, self.hx)
        return hidden


class TimedQNNOps(nn.Cell):
    def __init__(self, inner_ops, runtime_stats):
        super().__init__()
        self.inner_ops = inner_ops
        self.runtime_stats = runtime_stats

    def construct(self, x):
        start_time = time.perf_counter()
        output = self.inner_ops(x)
        elapsed = time.perf_counter() - start_time
        self.runtime_stats.qnn_forward_calls += 1
        self.runtime_stats.qnn_forward_seconds += elapsed
        return output


class All_Net(nn.Cell):
    def __init__(self, grad_ops, init_paras, n_parameters, runtime_stats):
        super().__init__()
        self.runtime_stats = runtime_stats
        self.lstm = TrainableLSTM(n_parameters)
        raw_qnn_ops = MQEncoderOnlyOps(grad_ops)
        self.qnn_ops = TimedQNNOps(raw_qnn_ops, runtime_stats)
        self.init_paras = init_paras
        self.reshape = ops.Reshape()

    def construct(self):
        lstm_start = time.perf_counter()
        lstm_output = self.lstm(self.init_paras)
        self.runtime_stats.lstm_forward_seconds += time.perf_counter() - lstm_start

        # ✅ 关键修复：必须保持二维形状 (batch_size=1, n_parameters)
        # MQEncoderOnlyOps 要求输入为 2D Tensor，shape=(batch, n_params)
        params = self.reshape(lstm_output, (1, -1))

        energy = self.qnn_ops(params)
        return energy


# ============================================================
# 5. 主流程
# ============================================================
if __name__ == "__main__":
    # --- 构建分子与 UCCSD 线路 ---
    ZrO2d = get_mol(1)
    active_space_start = 27
    active_space_stop = 32
    threshold = -1

    (ansatz_circuit, init_amplitudes, ansatz_parameter_names,
     hamiltonian_QubitOp, n_qubits, n_electrons) = generate_uccsd(
        ZrO2d, threshold=threshold,
        active_space_start=active_space_start,
        active_space_stop=active_space_stop,
    )

    hartreefock_wfn_circuit = Circuit([X.on(i) for i in range(n_electrons)])
    total_circuit = hartreefock_wfn_circuit + ansatz_circuit

    print("\n===== UCCSD ansatz circuit summary =====")
    total_circuit.summary()

    circ = hartreefock_wfn_circuit + ansatz_circuit.as_encoder()
    circuit_metrics = circuit_resource_metrics(circ)

    print("\n===== Static circuit resources =====")
    for k, v in circuit_metrics.items():
        print(f"  {k}: {v}")

    # --- 模拟器与哈密顿量 ---
    ham = Hamiltonian(hamiltonian_QubitOp)
    sim = Simulator("mqvector", circ.n_qubits)
    grad_ops = sim.get_expectation_with_grad(ham, circ)

    # --- 构建 LSTM-VQE 网络 ---
    n_parameters = len(circ.params_name)
    if n_parameters == 0:
        raise ValueError("No trainable parameters found in the circuit.")

    np.random.seed(0)
    initial_parameter_tensor = Tensor(np.random.random((1, 1, n_parameters)).astype(np.float32))
    runtime_stats = RuntimeStats()

    all_net = All_Net(
        grad_ops=grad_ops,
        init_paras=initial_parameter_tensor,
        n_parameters=n_parameters,
        runtime_stats=runtime_stats,
    )
    optimizer = nn.Adam(all_net.trainable_params(), learning_rate=0.1)
    train_net = nn.TrainOneStepCell(all_net, optimizer)
    train_net.set_train()

    # --- 训练循环 ---
    energy_data = []
    step_times = []
    steps = 50
    objective_forward_evaluations = 0
    gradient_calls = 0

    training_start = time.perf_counter()
    for step in range(steps):
        step_start = time.perf_counter()
        energy = train_net()
        energy_value = float(np.asarray(energy.asnumpy()).reshape(-1)[0])
        step_elapsed = time.perf_counter() - step_start

        energy_data.append(energy_value)
        step_times.append(step_elapsed)
        objective_forward_evaluations += 1
        gradient_calls += 1

        if step % 10 == 0:
            print(f"Step {step + 1}/{steps}: Energy = {energy_value:.12f} Hartree, time = {step_elapsed:.4f}s")

    total_training_wall_seconds = time.perf_counter() - training_start

    # --- 收敛检测 ---
    reference_energy = None
    convergence_tolerance = 1e-3
    converged_step = None
    if reference_energy is not None:
        for idx, e in enumerate(energy_data, start=1):
            if abs(e - reference_energy) <= convergence_tolerance:
                converged_step = idx
                break

    # --- 打印统计信息 ---
    print("\n===== Training resource statistics =====")
    print(f"Training steps: {len(energy_data)}")
    print(f"Objective forward evaluations: {objective_forward_evaluations}")
    print(f"Gradient calls: {gradient_calls}")
    print(f"QNN forward calls: {runtime_stats.qnn_forward_calls}")
    print(f"QNN forward wall time: {runtime_stats.qnn_forward_seconds:.6f}s")
    print(f"LSTM forward wall time: {runtime_stats.lstm_forward_seconds:.6f}s")
    print(f"Total training wall time: {total_training_wall_seconds:.6f}s")
    if step_times:
        print(f"Mean time per step: {np.mean(step_times):.6f}s")
    print(f"Final energy: {energy_data[-1]:.12f} Hartree")
    if converged_step is not None:
        print(f"Converged at step: {converged_step}")
    else:
        print("Convergence: not reached (set reference_energy to enable)")

    # --- 绘图 ---
    plt.figure()
    plt.plot(range(1, len(energy_data) + 1), energy_data, label="LSTM-VQE")
    plt.title("LSTM-VQE Training for Monoclinic ZrO2")
    plt.xlabel("Training step")
    plt.ylabel("Energy (Hartree)")
    plt.legend()
    plt.tight_layout()
    plt.savefig("lstm_vqe_zro2_energy.png", dpi=300)
    plt.show()

    # --- 保存 CSV ---
    csv_file = Path("resource_metrics.csv")
    resource_row = {
        "system": "Monoclinic ZrO2",
        "active_space_start": active_space_start,
        "active_space_stop": active_space_stop,
        "n_qubits": circuit_metrics["n_qubits"],
        "n_electrons": n_electrons,
        "n_parameters": circuit_metrics["n_parameters"],
        "gate_count": circuit_metrics["gate_count"],
        "parameterized_gate_count": circuit_metrics["parameterized_gate_count"],
        "estimated_circuit_depth": circuit_metrics["circuit_depth"],
        "training_steps": len(energy_data),
        "objective_forward_evaluations": objective_forward_evaluations,
        "gradient_calls": gradient_calls,
        "qnn_forward_calls": runtime_stats.qnn_forward_calls,
        "qnn_forward_seconds": runtime_stats.qnn_forward_seconds,
        "lstm_forward_seconds_only": runtime_stats.lstm_forward_seconds,
        "total_training_wall_seconds": total_training_wall_seconds,
        "converged_step": converged_step if converged_step is not None else "",
        "final_energy_hartree": energy_data[-1],
    }
    file_exists = csv_file.exists()
    with csv_file.open("a", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=list(resource_row.keys()))
        if not file_exists:
            writer.writeheader()
        writer.writerow(resource_row)
    print(f"\nResource metrics saved to: {csv_file.resolve()}")
```

```python
import numpy as np
import sys
from openfermion.chem import MolecularData
from openfermionpyscf import run_pyscf
from pyscf import mp  # 直接导入 pyscf 的 mp 模块

# ==============================================================================
# 1. 修复 NumPy 2.0 兼容性问题 (必须在 openfermion 底层调用前生效)
# ==============================================================================
if not hasattr(np, "string_"):
    np.string_ = np.bytes_
    sys.modules["numpy"].string_ = np.bytes_
# ==============================================================================


def get_noon_from_molecule(molecule):
    """
    从 MolecularData 对象中提取并分析 NOON
    基于底层已经收敛的 SCF 对象，直接调用 PySCF 的 MP2 模块。
    """
    # 检查是否存在 SCF 数据
    if not hasattr(molecule, '_pyscf_data') or 'scf' not in molecule._pyscf_data:
        print("❌ 错误: 未找到 SCF 数据，无法计算 MP2/NOON。请确保 run_pyscf 成功执行。")
        return

    print(f"\n正在基于 SCF 结果计算 MP2 以获取 NOON (通常只需几秒)...")
    
    # 获取底层跑好的 SCF 对象
    mf = molecule._pyscf_data['scf']
    
    # 利用现成的 mf 实例化 MP2 并运行
    pt = mp.MP2(mf)
    pt.kernel()
    
    # 获取 MP2 的 1-RDM (在分子轨道 MO 基下)
    rdm1 = pt.make_rdm1(ao_repr=False)
    
    # 假设闭壳层单重态，计算占据轨道数 (nocc)
    nocc = mf.mol.nelectron // 2
    nvirt = rdm1.shape[0] - nocc

    print(f"\n" + "="*45)
    print(f"🌟 NOON Analysis for {molecule.description} 🌟")
    print("="*45)
    print(f"{'Position':<12} | {'Global Index':<12} | {'NOON (n_i)':<10}")
    print("-" * 45)

    # 1. 处理占据轨道 (Occupied)
    rdm1_occ = rdm1[:nocc, :nocc]
    noon_occ, _ = np.linalg.eigh(rdm1_occ)
    idx_occ = np.argsort(noon_occ)[::-1]
    noon_occ = noon_occ[idx_occ]

    for i in range(min(5, nocc)): 
        pos_name = "HOMO" if i == 0 else f"HOMO-{i}"
        global_idx = nocc - 1 - i 
        print(f"{pos_name:<12} | {global_idx:<12} | {noon_occ[i]:.4f}")

    print("-" * 45)

    # 2. 处理虚轨道 (Virtual)
    rdm1_virt = rdm1[nocc:, nocc:]
    noon_virt, _ = np.linalg.eigh(rdm1_virt)
    idx_virt = np.argsort(noon_virt)[::-1]
    noon_virt = noon_virt[idx_virt]

    for i in range(min(5, nvirt)):
        pos_name = "LUMO" if i == 0 else f"LUMO+{i}"
        global_idx = nocc + i
        print(f"{pos_name:<12} | {global_idx:<12} | {noon_virt[i]:.4f}")
        
    print("-" * 45)
    
    # 3. 自动筛选符合 0.1 < n < 1.9 的轨道
    print("🎯 筛选结果 (满足 0.1 < n_i < 1.9):")
    print(f"{'Position':<12} | {'Global Index':<12} | {'NOON (n_i)':<10}")
    
    valid_orbitals = []
    
    for i in range(min(10, nocc)):
        if 0.1 < noon_occ[i] < 1.9:
             pos_name = "HOMO" if i == 0 else f"HOMO-{i}"
             global_idx = nocc - 1 - i
             print(f"  {pos_name:<10} | {global_idx:<12} | {noon_occ[i]:.4f}")
             valid_orbitals.append(global_idx)
             
    for i in range(min(10, nvirt)):
        if 0.1 < noon_virt[i] < 1.9:
             pos_name = "LUMO" if i == 0 else f"LUMO+{i}"
             global_idx = nocc + i
             print(f"  {pos_name:<10} | {global_idx:<12} | {noon_virt[i]:.4f}")
             valid_orbitals.append(global_idx)
             
    if not valid_orbitals:
        print("  (没有轨道满足筛选条件)")
    else:
        print(f"\n👉 建议的活性空间轨道索引 (供参考): {sorted(valid_orbitals)}")
    print("="*45 + "\n")


def get_mol(phase=1):
    # 根据输入phase选择不同的晶体结构
    if phase == 1:
        # ① 单斜相ZrO2
        a = 5.15; b = 5.23; c = 5.33
        beta = 99.44 * np.pi / 180
        a_vec = np.array([a, 0, 0])
        b_vec = np.array([0, b, 0])
        c_vec = np.array([c * np.cos(beta), 0, c * np.sin(beta)])
        atoms = [
            ('Zr', [0.276309, 0.042147, 0.209598]),
            ('O',  [0.930791, 0.833712, 0.156172]),
            ('O',  [0.449303, 0.742580, 0.978604])
        ]
        description = "Monoclinic ZrO2"
    elif phase == 2:
        # ② 四方相ZrO2
        a = b = 3.60; c = 5.23
        a_vec = np.array([a, 0, 0]); b_vec = np.array([0, b, 0]); c_vec = np.array([0, 0, c])
        atoms = [
            ('Zr', [0, 0, 0]),
            ('O',  [0, 0.5, 0.301775])
        ]
        description = "Tetragonal ZrO2"
    elif phase == 3:
        # ③ 六方相ZrO
        a = 5.31; c = 3.2
        a_vec = np.array([a, 0, 0])
        b_vec = np.array([-a/2, a*np.sqrt(3)/2, 0])
        c_vec = np.array([0, 0, c])
        atoms = [
            ('Zr', [0, 0, 0]),
            ('O',  [1/3, 2/3, 0.25])
        ]
        description = "Hexagonal ZrO"
    else:
        raise ValueError("Invalid phase. Choose 1, 2, or 3.")

    # 将分数坐标转换为笛卡尔坐标
    geometry = []
    for element, frac_coords in atoms:
        cart_coords = (
            frac_coords[0] * a_vec +
            frac_coords[1] * b_vec +
            frac_coords[2] * c_vec
        )
        geometry.append([element, list(cart_coords)])

    basis = 'gth-dzvp-molopt-sr'
    multiplicity = 1  # 假设为闭壳层单重态

    # 创建分子对象
    molecule = MolecularData(
        geometry,
        basis,
        multiplicity,
        description=description
    )

    # 运行或加载计算
    try:
        molecule.load()
        # 检查核心数据是否齐全，不齐全则触发重算
        if not hasattr(molecule, '_pyscf_data') or 'scf' not in molecule._pyscf_data:
            raise FileNotFoundError("SCF data missing in cache.")
    except:
        print(f"Running HF and CCSD calculations for {description}... (请耐心等待)")
        # 注意：因为我们要手动跑 MP2，所以这里不需要 run_mp2=1 了
        molecule = run_pyscf(
            molecule,
            run_scf=1,
            run_ccsd=1,
        )
        molecule.save()
        molecule.load()
        
    # 打印能量信息
    print(f'\n[{description}]')
    print(f'  HF energy:   {molecule.hf_energy:.6f} Hartree')
    print(f'  CCSD energy: {molecule.ccsd_energy:.6f} Hartree')
    
    # 调用 NOON 提取函数 (内部会自动快速算一次 MP2)
    get_noon_from_molecule(molecule)
    
    return molecule

if __name__ == "__main__":
    print(">>> 开始计算单斜相 (Monoclinic ZrO2)...")
    mol1 = get_mol(phase=1)
    
    print("\n>>> 开始计算四方相 (Tetragonal ZrO2)...")
    mol2 = get_mol(phase=2)
    
    print("\n>>> 开始计算六方相 (Hexagonal ZrO)...")
    mol3 = get_mol(phase=3)
    
    print("\n✅ 所有计算完成！请将上方打印的 NOON 数据填入您的 Table 2。")
```
