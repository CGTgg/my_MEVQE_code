## 能量修正后 与 HF及CCSD方法对比

```python
import matplotlib.pyplot as plt

# 数据
methods = ["Hartree-Fock", "CCSD", "MEVQE"]
energies = [-608.1196082141597, -608.3663660904948, -608.3679162169106]  # Hartree

# 创建图形
plt.figure(figsize=(10, 6))

# 标记形状和颜色配置
markers = ['o', 's', '^']  # 圆形、正方形、三角形、五角星
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
marker_sizes = [80, 80, 80]  # 五角星稍大以提高辨识度

# 绘制散点图（每个方法一个点）
# 绘制散点图（不同形状和颜色）
for i, (method, energy) in enumerate(zip(methods, energies)):
    plt.scatter(method, energy, 
                marker=markers[i], 
                color=colors[i], 
                s=marker_sizes[i], 
                label=method, 
                #edgecolor='black',  # 标记边框
                linewidth=0.8,      # 边框粗细
                zorder=3)

# 添加水平参考线（浅灰色虚线）
#for energy in energies:
#    plt.axhline(y=energy, color='gray', linestyle='-.', alpha=0.2, zorder=0.1)

# 坐标轴标签和标题
plt.xlabel("Method", fontsize=12)
plt.ylabel("Energy (Hartree)", fontsize=12)
plt.title("Monoclinic ZrO₂ Energy Comparison", fontsize=14, pad=20)

# 调整纵坐标范围以突出差异
plt.ylim(min(energies) - 0.1, max(energies) + 0.1)

# 网格线（横向）
plt.grid(axis='y', linestyle='--', alpha=0.5)

# 显示图例（可选）
plt.legend(loc='upper right', framealpha=1)

plt.tight_layout()
#plt.savefig("./fig/Menergy_comparison.png", dpi=300)  # 保存高清图
plt.show()
```
